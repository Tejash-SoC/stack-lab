```python
# 1. Task Scheduling Problem 
# Write an asyncio program where two API calls take different 
# times to complete. Use asyncio.create_task() so both start 
# immediately, and print which task finishes first.  

import asyncio

async def api1():
    print("API 1 Started")
    await asyncio.sleep(3)
    print("API 1 Completed")
    return "API 1"

async def api2():
    print("API 2 Started")
    await asyncio.sleep(2)
    print("API 2 Completed")
    return "API 2"

async def main():

    # Create tasks (both start immediately)
    task1 = asyncio.create_task(api1())
    task2 = asyncio.create_task(api2())

    # Wait until the first task completes
    done, pending = await asyncio.wait(
        [task1, task2],
        return_when=asyncio.FIRST_COMPLETED
    )

    # Print the first completed task
    for task in done:
        print(f"\nFirst Finished: {task.result()}")

    # Wait for the remaining task to finish
    for i in pending:
        i.cancel()
    # await asyncio.gather(*pending)

await main()
```

```python
# 4. Wait with Timeout Problem 
# Create three tasks simulating file downloads: 1 sec, 3 sec, and 5 
# sec. Use asyncio.wait() with a timeout of 2 seconds. Print 
# completed downloads and cancel pending ones. 


import asyncio
import time

async def c1():
    print("c1 Started")
    await asyncio.sleep(1)
    print("c1 Completed")
    return "c1"

async def c2():
    print("c2 Started")
    await asyncio.sleep(3)
    print("c2 Completed")
    return "c2"

async def c3():
    print("c3 Started")
    await asyncio.sleep(5)
    print("c3 Completed")
    return "c3"

async def main():
    t1=asyncio.create_task(c1())
    t2=asyncio.create_task(c2())
    t3=asyncio.create_task(c3())
    d,p=await asyncio.wait([t1,t2,t3],timeout=2)
    for i in d:
        print(i.result())
    for i in p:
        i.cancel()
await main()
```

```python
# 6. Timeout Protection Problem 
# Write a coroutine that simulates a slow database query. Use 
# asyncio.wait_for() to cancel it if it takes more than 2 seconds, 
# and print "Database timeout".  

import asyncio
import time

async def c1():
    print("c1 Started")
    await asyncio.sleep(5)
    print("c1 Completed")
    return "c1"
async def main():
    try:
        await asyncio.wait_for(c1(),timeout=2)
    # except Exception as e:
    #     print(type(e).__name__)
    except asyncio.TimeoutError:
        print("Database timeout")

await main()
```

```python
# 7. Cancellation Problem 
# Create a long-running background task like sending email 
# notifications. Cancel it after 2 seconds and handle 
# asyncio.CancelledError properly.  

import asyncio
import time

async def c1():
    print("c1 Started")
    await asyncio.sleep(5)
    print("c1 Completed")
    return "c1"
async def main():
    t1=asyncio.create_task(c1())
    
    await asyncio.sleep(2)
    t1.cancel()
    try:
        await t1
    except asyncio.CancelledError:
        print("Cancelled")
    
    
await main()
```

```python
# 9. Lock Synchronization Problem 
# Create three coroutines trying to update the same shared account 
# balance. Use asyncio.Lock() so only one coroutine updates the 
# balance at a time.  

import asyncio

balance = 100  # shared account balance
lock = asyncio.Lock()

async def update_balance(name, amount):
    print(f"{name} wants to update balance...")
    async with lock:  # acquire lock
        print(f"{name} acquired lock")
        global balance
        old_balance = balance
        await asyncio.sleep(1)  # simulate some processing
        balance = old_balance + amount
        print(f"{name} updated balance to {balance}")
    print(f"{name} released lock")

async def main():
    # Three coroutines trying to update the same balance
    t1 = asyncio.create_task(update_balance("Task1", 50))
    t2 = asyncio.create_task(update_balance("Task2", -30))
    t3 = asyncio.create_task(update_balance("Task3", 20))

    await asyncio.gather(t1, t2, t3)
    print("Final balance:", balance)

await main()
```

```python
# 10. 
# Producer–Consumer Queue Problem 
# Build an async producer that adds 10 jobs into an 
# asyncio.Queue, and two consumers that process the jobs 
# concurrently. Use queue.join() and cancel consumers after all 
# jobs are done. 



import asyncio

async def producer(queue):
    for i in range(10):
        await asyncio.sleep(0.2)  # simulate job creation
        job = f"Job-{i}"
        await queue.put(job)
    #     print(f"Produced {job}")
    # print("Producer finished adding jobs")

async def consumer(name, queue):
    try:
        while True:
            job = await queue.get()
            print(f"{name} processing {job}")
            await asyncio.sleep(1)  # simulate work
            queue.task_done()
    except asyncio.CancelledError:
        print(f"{name} cancelled")
        raise

async def main():
    queue = asyncio.Queue()
    
    consumers = [      # Start consumers
        asyncio.create_task(consumer("Consumer-1", queue)),
        asyncio.create_task(consumer("Consumer-2", queue)),
    ] 
    
    await producer(queue) # Start producer

    await queue.join() # Wait until all jobs are processed

    for c in consumers: # Cancel consumers after all jobs are done
        c.cancel()

    await asyncio.gather(*consumers, return_exceptions=True) # Ensure cancellation is handled
    print("All jobs processed, consumers cancelled")

await main()
    
```
