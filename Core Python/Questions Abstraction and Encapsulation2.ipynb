```python
# 1.  Create a BankAccount class that stores: 
# • account number 
# • balance (should not be directly modifiable) 
# You must: 
# 1. Make the balance attribute inaccessible from outside. 
# 2. Provide functions to deposit/withdraw that validate the amount. 
# 3. Prevent withdrawal if balance becomes negative. 
# 4. Show what happens if someone tries to modify balance directly and why 
# encapsulation prevents it. 

class BankAccount:
    def __init__(self,account_number,balance):
        self.account_number=account_number
        self.__balance=balance
    def show(self):
        return self.__balance
    def deposit(self,amount):
        self.__balance+=amount
    def withdraw(self,amount):
        if(amount<0):
            print("Amount is negative")
            return
        if(amount>self.__balance):
            print("Invalid amount")
        else:
            self.__balance-=amount
a1=BankAccount(123,100)
a1._BankAccount__balance
a1.deposit(20)
a1.show()
a1.withdraw(1)
a1.show()
```

```python
# 3. Create a SecureFile class that: 
# • stores content privately 
# • provides a method read(password) 
# • refuses access if the password is incorrect 
# • logs an "Unauthorized attempt" internally (cannot be accessed from outside)

class SecureFile:
    def __init__(self,content):
        self.__content=content
    def read(self,password):
        if(password==1234):
            return self.__content
        else:
            print("Unauthorized attempt")
a1=SecureFile('Hello')
a1.read(1234)
```

```python
# 5. Create a Product class where: 
# • price cannot be negative 
# • discount cannot exceed 70% 
# • internal final price calculation should not be directly exposed 
# Provide only one public method get_final_price().

class Product:
    def __init__(self,price,discount):
        if(price<0):
            raise ValueError
        if(0>discount or discount>70):
            raise ValueError
        self.price=price
        self.discount=discount
    def get_final_price(self):
        __finalprice=self.price-(self.price*self.discount/100)
        return __finalprice
a1=Product(100,10)
a1.get_final_price()
```

```python
# 6. Create a Character class with: 
# • private _health 
# • methods to damage(points) and heal(points) 
# • health cannot drop below 0 or exceed max limit 
# • expose only current health through a read-only getter 


class Character:
    def __init__(self):
        self.__health=100
    def damage(self,a):
        if((self.__health-a)>=0):
            self.__health-=a
        else:
            print("Invalid damage")
    def heal(self,a):
        if((self.__health+a)<=100):
            self.__health=+a
        else:
            print("Invalid heal")
    def get(self):
        return self.__health
a1=Character()
print(a1.get())
a1.heal(10)
a1.damage(110)
a1.damage(29)
print(a1.get())

```

```python
# 7. Create: 
# • An Engine class with private state like temperature 
# • A Car class that uses an Engine but should: 
# o Not allow users to manipulate engine temperature 
# o Only expose methods like start_car() or cool_engine()

class Engine:
    def __init__(self):
        self._temperature = 25   # private state

    def _increase_temp(self, amount):
        self._temperature += amount

    def _cool_down(self, amount):
        self._temperature = max(0, self._temperature - amount)

    def _get_temperature(self):
        return self._temperature


class Car:
    def __init__(self):
        self._engine = Engine()

    def start_car(self):
        """Start car → engine heats up internally."""
        self._engine._increase_temp(50)
        print("Car started. Engine temperature:", self._engine._get_temperature())

    def cool_engine(self):
        """Cool engine safely without direct access."""
        self._engine._cool_down(30)
        print("Engine cooled. Temperature:", self._engine._get_temperature())
mycar = Car()
mycar.start_car()    # Car started. Engine temperature: 75
mycar.cool_engine()  # Engine cooled. Temperature: 45


```

```python
# 8. Create a ShoppingCart class where: 
# • items are stored privately 
# • users cannot directly modify item list 
# • only add/remove methods are allowed 
# • provide a method to get a safe copy of the cart items (not direct reference to internal 
# list)

class ShoppingCart:
    def __init__(self):
        self.__items=[]
    def add(self,i):
        self.__items.append(i)
    def remove(self,i):
        self.__items.remove(i)
    def c(self):
        return self.__items
a1=ShoppingCart()
print(a1.c())
a1.add('cool drink')
a1.add('Biscuit')
print(a1.c())
a1.remove('Biscuit')
print(a1.c())


```
