```python
# • Create a class BankAccount with an attribute balance. Implement a method 
# withdraw(amount) that raises an exception if the withdrawal amount is greater 
# than the available balance.


class Bankaccount:
    def __init__(self,balance):
        self.balance=balance
    def withdraw(self,amount):
        if(amount>self.balance):
            raise ValueError("self.withdrawal amount is greater then balance")
        else:
            self.balance-=amount
a1=Bankaccount(1000)
a1.withdraw(20)
a1.withdraw(1000)
```

```python
# • Create a class Transaction with a method process() that uses try, except, and 
# finally blocks to ensure a cleanup message is always printed.

class Transaction:
    def process(self,a,b):
        try:
            k=a/b
        except TypeError:
            print("Type should be integer")
        except ZeroDivisionError:
            print("Denominator should not zero")
        else:
            print(k)
        finally:
            print("Finished")
a1=Transaction()
a1.process(2,3)
a1.process(2,0)
a1.process(2,'2')
```
