```python
# 11. Using abc module: 
# • Create an abstract class Shape with area(), perimeter() 
# • Implement Circle, Rectangle, Triangle 
# Demonstrate: 
# • why base class should NOT contain calculation logic 
# • what happens if a subclass fails to implement one of the methods

from abc import ABC,abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
    @abstractmethod
    def perimeter(self):
        pass
class Circle(Shape):
    def __init__(self,r):
        self.r=r
    def area(self):
        return 3.14*self.r*self.r
    def perimeter(self):
        return 2*3.14*self.r
class Rectangle(Shape):
    def __init__(self,l,b):
        self.l=l
        self.b=b
    def area(self):
        return self.l*self.b
    def perimeter(self):
        return 2*self.l*self.b
class Triangle(Shape):
    def __init__(self,h,b):
        self.h=h
        self.b=b
    def area(self):
        return (1/2)*self.h*self.b
    def perimeter(self):
        return self.h*self.b
a1=Rectangle(2,3)
print(a1.area())
print(a1.perimeter())
    
```

```python
# 12. Design an abstract class PaymentGateway with: 
# • authenticate() 
# • pay(amount) 
# • refund(amount) 
# Implement subclasses: 
# • UPIPayment 
# • CardPayment 
# • NetBankingPayment 
# Show how abstraction helps your main program call payment methods without caring about 
# the payment type. 


from abc import ABC,abstractmethod
class PaymentGateway(ABC):
    @abstractmethod
    def authenticate(self):
        pass
    @abstractmethod
    def pay(self,amount):
        pass
    @abstractmethod
    def refund(self,amount):
        pass
class UPIPayment(PaymentGateway):
    def authenticate(self):
        print('authenticate in UPIPayment')
    def pay(self, amount):
        print(f'pay {amount} in UPIPayment')
    def refund(self, amount):
        print(f'refund {amount} in UPIPayment')
class CardPayment(PaymentGateway):
    def authenticate(self):
        print('authenticate in CardPayment')
    def pay(self, amount):
        print(f'pay {amount} in CardPayment')
    def refund(self, amount):
        print(f'refund {amount} in CardPayment')
class NetBankingPayment(PaymentGateway):
    def authenticate(self):
        print('authenticate in NetBankingPayment')
    def pay(self, amount):
        print(f'pay {amount} in NetBankingPayment')
    def refund(self, amount):
        print(f'refund {amount} in NetBankingPayment')
a1=UPIPayment()
a1.pay(3)
a2=CardPayment()
a2.refund(3)
a3=NetBankingPayment()
a3.authenticate()

```

```python
# 13. Create: 
# • Abstract class VehicleControl with methods accelerate(), brake(), steer() 
# • Implement CarControl, BikeControl, TruckControl 
# Demonstrate calling each through a single interface. 


from  abc import ABC,abstractmethod
class VehicleControl(ABC):
    @abstractmethod
    def accelerate(self):
        pass
    @abstractmethod
    def brake(self):
        pass
    @abstractmethod
    def steer(self):
        pass
class CarControl(VehicleControl):
    def accelerate(self):
        print("Accelerting Car")
    def brake(self):
        print("Breaking Car")
    def steer(self):
        print("Steering Car")
class BikeControl(VehicleControl):
    def accelerate(self):
        print("Accelerting Bike")
    def brake(self):
        print("Breaking Bike")
    def steer(self):
        print("Steering Bike")
class TruckControl(VehicleControl):
    def accelerate(self):
        print("Accelerting Truck")
    def brake(self):
        print("Breaking Truck")
    def steer(self):
        print("Steering Truck")
a1=CarControl()
a1.accelerate()
a2=BikeControl()
a2.brake()
a3=TruckControl()
a3.steer()
```

```python
# 14. Create an abstract class DatabaseDriver with: 
# • connect() 
# • execute(query) 
# • close() 
# Implement concrete drivers: 
# • MySQLDriver 
# • PostgresDriver 
# • SQLiteDriver 
# Show how abstraction helps switch databases without rewriting main code. 


from  abc import ABC,abstractmethod
class DatabaseDriver(ABC):
    @abstractmethod
    def connect(self):
        pass
    @abstractmethod
    def execute(self,query):
        pass
    @abstractmethod
    def close(close):
        pass
class MySQLDriver(DatabaseDriver):
    def connect(self):
        print("Connecting MySQLDriver")
    def execute(self, query):
        print(f"executing {query} MySQLDriver")
    def close(self):
        print("Closing MySQLDriver")
class PostgresDriver(DatabaseDriver):
    def connect(self):
        print("Connecting PostgresDriver")
    def execute(self, query):
        print(f"executing {query} PostgresDriver")
    def close(self):
        print("Closing PostgresDriver")
class SQLiteDriver(DatabaseDriver):
    def connect(self):
        print("Connecting SQLiteDriver")
    def execute(self, query):
        print(f"executing {query} SQLiteDriver")
    def close(self):
        print("Closing SQLiteDriver")
a1=MySQLDriver()
a1.connect()
a2=PostgresDriver()
a2.execute(23)
a3=SQLiteDriver()
a3.close()
```

```python
# 16. Create an abstract class RobotCommand with: 
# • execute() 
# • undo() 
# Implement: 
# • PickCommand 
# • PlaceCommand 
# • MoveCommand 
# Demonstrate how abstraction cleanly represents commands without revealing details. 

from abc import ABC,abstractmethod
class RobotCommand(ABC):
    @abstractmethod
    def execute(self):
        pass
    @abstractmethod
    def undo(self):
        pass
class PickCommand(RobotCommand):
    def execute(self):
        print("Executing PickCommand")
    def undo(self):
        print("undo PickCommand")
class PlaceCommand(RobotCommand):
    def execute(self):
        print("Executing PlaceCommand")
    def undo(self):
        print("undo PlaceCommand")
class MoveCommand(RobotCommand):
    def execute(self):
        print("Executing MoveCommand")
    def undo(self):
        print("undo MoveCommand")
l=[PickCommand(), PlaceCommand(), MoveCommand()]
for i in l:
    i.execute()
    i.undo()
```

```python
# 17. Create an abstract class MLModel with: 
# • train(data) 
# • predict(x) 
# • evaluate(test_set) 
# Implement models: 
# • LinearRegressionModel- some different logic 
# • DecisionTreeModel – some logic 
# Show how a generic training loop works for any model without caring about details. 


from abc import ABC,abstractmethod
class MLModel(ABC):
    @abstractmethod
    def train(self,data):
        pass
    @abstractmethod
    def predict(self,x):
        pass
    @abstractmethod
    def evaluate(self,test_set):
        pass
class LinearRegressionModel(MLModel):
    def train(self,data):
        print(f"Training {data} using LinearRegressionModel")
    def predict(self,x):
        print(f"predict {x} using LinearRegressionModel")
    def evaluate(self,test_set):
        print(f"evaluate {test_set} using LinearRegressionModel")
class DecisionTreeModel(MLModel):
    def train(self,data):
        print(f"Training {data} using DecisionTreeModel")
    def predict(self,x):
        print(f"predict {x} using DecisionTreeModel")
    def evaluate(self,test_set):
        print(f"evaluate {test_set} using DecisionTreeModel")

data = [1, 2, 3]
x = [4, 5]
test_set = [6, 7, 8]

l = [LinearRegressionModel(), DecisionTreeModel()]
for i in l:
    i.train(data)
    i.predict(x)
    i.evaluate(test_set)
```
