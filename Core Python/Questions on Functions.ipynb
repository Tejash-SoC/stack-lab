```python
# 1. Write a function that accepts a list and returns the sum of even numbers.

l=list(map(int,input().split()))
def en():
    s=0
    for i in l:
        if(i%2==0):
            s+=i
    return s
print(en())
```

```python
# 2. Create a function that takes a string and returns the number of vowels.

s=input()
def nv():
    v=0
    for i in s:
        if i in "aeiouAEIOU":
            v=v+1
    return v
print(nv())
```

```python
# 3. Write a function using **kwargs to print student details.

def f(**kwargs):
    for k,v in kwargs.items():
        print(f"{k}:{v}")
f(a=1,b=2,c=3)
```

```python
# 4. Write a function where inner function modifies outer variable using nonlocal
def out():
    a=1
    print(a)
    def inn():
        nonlocal a
        a=2
        print(a)
    return inn()
out()
```

```python
# 5. Convert list of temperatures in Celsius to Fahrenheit using map().

l=[1,2,3,4,5,6]
fah=map(lambda x:x*9/5+32,l)
print(list(fah))
```

```python
# 7. Use map() with lambda to extract first letters of names.

l=["hi","hello","bye"]
k=list(map(lambda x:x[0],l))
print(k)
```

```python
# 8. Filter strings that start with vowel.

l=['ashok','reddy','chitti']
print(list(filter(lambda x:x[0] in "aeiouAEIOU",l)))
```

```python
# 9. Multiply only even numbers using reduce().

from functools import reduce
l=map(int,input().split())
en=list(filter(lambda x:x%2==0,l))
print(reduce(lambda x,y:x*y,en))
```

```python
# 10. Compute factorial of a number using reduce().

n=int(input())
print(reduce(lambda x,y:x*y,range(1,n+1)))
```

```python
# 11. Use filter() + reduce() to sum only odd numbers.

l=map(int,input().split())
k=list(filter(lambda x:x%2==1,l))
print(reduce(lambda x,y:x+y,k))
```
