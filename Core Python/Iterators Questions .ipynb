```python
# 1. Write a custom iterator that prints numbers from 1 to N.
class A:
    def __init__(self,n):
        self.n=n
        self.c=0
    def __iter__(self):
        return self
    def __next__(self):
        self.c+=1
        if(self.c<=self.n):
            return self.c
        else:
            raise StopIteration
n=int(input())
a1=A(n)
for i in a1:
    print(i)
```

```python
# 6. Create a generator that yields cumulative sum of numbers in a list. Example: 
# [1,2,3] → 1, 3, 6

def A(l):
    s=0
    for i in l:
        s+=i
        yield s
for i in A([1,2,3,4,5]):
    print(i)
```

```python
# 7. Implement a generator that yields vowels from a string.

def a(s):
    v='a', 'e', 'i', 'o', 'u'
    for i in s:
        if(i in v):
            yield i
for i in a('Tejash'):
    print(i)
```

```python
# 10. Implement a generator that yields running maximum from a list Example: 
# [3,1,4,2] → 3, 3, 4, 4

def a(l):
    m=0
    for i in l:
        if(i>m):
            m=i
        yield m
for i in a([4,2,6,1,8,5]):
    print(i)
```
